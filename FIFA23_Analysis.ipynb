# FIFA 23 Players Data Analysis

## Project Overview
This project analyzes FIFA 23 player data using Python to explore player ratings, potential, growth, value, age, positions, and clubs.

## Analysis Steps
- Load and inspect the dataset
- Check missing values and duplicate records
- Clean the dataset
- Explore player ratings and potential
- Calculate player growth
- Analyze correlations between key variables
- Analyze player positions and clubs
- Explore player value
- Visualize the distribution of player ratings

## Tools Used
- Python
- Pandas
- Matplotlib
- Jupyter Notebook

```python
import pandas as pd
```

```python
df = pd.read_csv("FIFA23.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.nlargest(10, 'Overall')[['Full Name', 'Overall', 'Potential', 'Value(in Euro)', 'Club Name']]
```

```python
df.nlargest(10, 'Potential')[['Full Name', 'Age', 'Overall', 'Potential', 'Value(in Euro)', 'Club Name']]
```

```python
df['Growth'] = df['Potential'] - df['Overall']
```

```python
df[['Full Name', 'Overall', 'Potential', 'Growth']].sort_values('Growth', ascending=False).head(10)
```

```python
print("Missing values:")
print(df.isnull().sum().sort_values(ascending=False).head(15))

print("\nDuplicate rows:")
print(df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Rows after removing duplicates:", len(df))
```

```python
df[['Overall', 'Potential', 'Growth']].corr()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.scatter(df['Overall'], df['Potential'], alpha=0.3)

plt.xlabel('Overall Rating')
plt.ylabel('Potential')
plt.title('Overall vs Potential')

plt.show()
```

```python
top_growth = df.sort_values('Growth', ascending=False).head(10)

top_growth[['Full Name', 'Age', 'Overall', 'Potential', 'Growth', 'Value(in Euro)', 'Club Name']]
```

```python
df[['Age', 'Overall', 'Potential', 'Growth']].corr()
```

```python
df[['Overall', 'Potential', 'Value(in Euro)']].corr()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.scatter(df['Overall'], df['Value(in Euro)'], alpha=0.3)
plt.xlabel('Overall Rating')
plt.ylabel('Value (Euro)')
plt.title('Overall Rating vs Player Value')
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df['Age'], df['Value(in Euro)'], alpha=0.3)

plt.xlabel('Age')
plt.ylabel('Value (Euro)')
plt.title('Age vs Player Value')

plt.show()
```

```python
position_analysis = df.groupby('Best Position')['Overall'].agg(['count', 'mean']).sort_values('count', ascending=False)

position_analysis
```

```python
top_positions = position_analysis.head(10)

top_positions['count'].sort_values().plot(
    kind='barh',
    figsize=(8,5),
    title='Top 10 Positions by Number of Players'
)

plt.xlabel('Number of Players')
plt.ylabel('Position')
plt.show()
```

```python
club_counts = df['Club Name'].value_counts().head(10)

club_counts.sort_values().plot(
    kind='barh',
    figsize=(8,5),
    title='Top 10 Clubs by Number of Players'
)

plt.xlabel('Number of Players')
plt.ylabel('Club')
plt.show()
```

```python
top_value = df.nlargest(10, 'Value(in Euro)')

top_value[['Full Name', 'Age', 'Overall', 'Potential',
           'Value(in Euro)', 'Wage(in Euro)', 'Club Name']]
```

```python
plt.figure(figsize=(8,5))

plt.hist(df['Overall'], bins=20, edgecolor='black')

plt.xlabel('Overall Rating')
plt.ylabel('Number of Players')
plt.title('Distribution of Player Overall Ratings')

plt.show()
```

```python
df.to_csv("FIFA23_Cleaned.csv", index=False)
```
